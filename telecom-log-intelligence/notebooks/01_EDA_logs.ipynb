```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_parquet("../data/processed/ml_dataset.parquet")

print("Dataset shape:", df.shape)

df.head()
```

```python
df.columns
```

```python
df.dtypes
```

```python
df.describe()
```

```python
df["platform"].value_counts().plot(kind="bar", figsize=(8,4))

plt.title("Distribution des plateformes")

plt.show()
```

```python
df["event_type"].value_counts().plot(kind="bar")

plt.title("Event types")

plt.show()
```

```python
df["message_length"] = df["message"].astype(str).apply(len)

sns.histplot(df["message_length"], bins=50)

plt.title("Distribution message length")

plt.show()
```

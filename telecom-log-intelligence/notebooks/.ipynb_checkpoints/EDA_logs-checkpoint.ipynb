```python
import pandas as pd
import numpy as np
from pymongo import MongoClient

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
client = MongoClient("mongodb://localhost:27017")

db = client["telecom_logs"]

collection = db["log_features"]
```

```python
cursor = collection.aggregate([
    {"$sample": {"size": 100000}}
])

df = pd.DataFrame(list(cursor))

df.head()
```

```python
df.info()
```

```python
df["platform"].value_counts()
```

```python
df["is_error"].value_counts()
```

```python
logs_per_hour = df.groupby("hour").size()

logs_per_hour.plot(kind="bar", figsize=(10,5))

plt.title("Logs Distribution by Hour")

plt.show()
```

```python
sns.histplot(df["message_length"], bins=50)

plt.title("Message Length Distribution")

plt.show()
```

```python
numeric = df.select_dtypes(include=["int64","float64","bool"])

corr = numeric.corr()

plt.figure(figsize=(10,8))

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Feature Correlation")

plt.show()
```
